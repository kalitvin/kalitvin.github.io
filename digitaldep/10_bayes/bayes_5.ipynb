Наивный Байес. Спам-фильтр.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Синтетический набор сообщений
texts = [
    # spam
    "Поздравляем вы выиграли приз заберите по ссылке",
    "Срочно получи кредит без справок и поручителей жми сюда",
    "Уникальное предложение только сегодня скидка 90 процентов",
    "Выбраны для участия в акции перейдите по ссылке",
    "Получите быстрый займ за 5 минут ответ онлайн",
    "Купи дешево брендовые товары переходи на сайт",

    # ham
    "Привет давай созвонимся вечером по проекту",
    "Отправляю отчет по продажам за месяц",
    "Напоминаю о встрече завтра в 15 часов",
    "Спасибо за помощь с документами все получилось",
    "Можем перенести обсуждение на следующий понедельник",
    "Файлы с презентацией прикрепил посмотри пожалуйста",
]

labels = [
    "spam", "spam", "spam", "spam", "spam", "spam",
    "ham", "ham", "ham", "ham", "ham", "ham"
]

# 2. Векторизация текстов (мешок слов)
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

# 3. Обучение наивного Байеса
clf = MultinomialNB()
clf.fit(X, labels)

# 4. Проверка на новых сообщениях
new_messages = [
    "Вы выиграли новый айфон заберите приз по ссылке",
    "Добрый день высылаю обновленный отчет по проекту",
    "Скидка на кредит только сегодня жмите сюда",
]

X_new = vectorizer.transform(new_messages)
pred = clf.predict(X_new)
proba = clf.predict_proba(X_new)

for msg, label, p in zip(new_messages, pred, proba):
    print("Сообщение:", msg)
    print("Класс:", label)
    print("P(ham), P(spam):", p)
    print("---")
```
