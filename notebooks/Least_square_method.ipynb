Метод наименьших квадратов (Least square method)

Пусть функция $f(x)$ задана таблицей
\begin{equation*}
\begin{array}{|c|c|c|c|c|}
\hline
x & x_{0} & x_{1} & \dots & x_{n}\\
\hline
f(x) & y_{0} & y_{1} & \dots & y_{n}\\
\hline
\end{array}
\end{equation*}

При нахождении аналитического выражения приближающей функции методом наименьших квадратов не требуется совпадения значений исходной и приближающей функций в точках $x_i.$ Однако необходимо учитывать класс функций, к которому относится искомая функция. На практике это можно сделать, отметив на графике исходные точки и учитывая дополнительные условия задачи. Часто используют следующие функции:
$$
\begin{array}{l}
1).\;y = ax + b \hbox{ (линейная функция)};\\
2).\;y = ax^{2} + bx + c \hbox{ (квадратичная функция)};\\
3).\;y = ax^{m} \hbox{ (степенная функция)};\\
4).\;y = ae^{mx} \hbox{ (показательная функция)};\\
5).\;y = \frac{{ax+b}}{{cx + d}} \hbox{ (дробно-линейная функция)};\\
6).\;y = a \cdot lnx + b \hbox{ (логарифмическая функция)};\\
7).\;y = \frac{{a}}{{x}} + b \hbox{ (гипербола)}.\\
\end{array}
$$

Для нахождения параметров конкретной приближающей формулы в методе наименьших квадратов требуется, чтобы сумма квадратов отклонений значений приближающей функции от значений исходной функции была минимальной.

Будем рассматривать нахождение параметров $a_1,a_2,\ldots ,a_n$ приближающей функции $F(x,a_1,a_2,\ldots ,a_n).$

Пусть
$$
\sum\limits_{i=0}^n \left(y_i-F(x_i,a_1,a_2,\ldots ,a_n)\right)^2 =\Phi \left( a_1,a_2,\ldots ,a_n \right).
$$

При применении метода наименьших квадратов необходимо найти точки минимума функции
$\Phi\!(a_1,a_2,\ldots ,a_n),$ т.е. точки, в которых

$$
\frac{\partial \Phi}{\partial a_1}=0,
\frac{\partial \Phi}{\partial a_2}=0,
\ldots ,
\frac{\partial \Phi}{\partial a_n}=0.
$$

Таким образом, получим систему $n$ линейных алгебраических уравнений с $n$ неизвестными
$$
\left\{
\begin{array}{l}
\sum\limits_{i=0}^n \left( y_i-F(x_i,a_1,a_2,\ldots ,a_n) \right)
\cdot \frac{\partial F(x_i,a_1,a_2,\ldots ,a_n)}{\partial a_1}=0,\\
\sum\limits_{i=0}^n \left( y_i-F(x_i,a_1,a_2,\ldots ,a_n) \right)
\cdot \frac{\partial F(x_i,a_1,a_2,\ldots ,a_n)}{\partial a_2}=0,\\
\ldots\\
\sum\limits_{i=0}^n \left( y_i-F(x_i,a_1,a_2,\ldots ,a_n) \right)
\cdot \frac{\partial F(x_i,a_1,a_2,\ldots ,a_n)}{\partial a_n}=0.\\
\end{array}
\right. (1)
$$

Решая эту систему, найдем значения параметров $a_1,a_2,\ldots ,a_n.$

Пример. Используя метод наименьших квадратов, найти приближающую функцию в виде линейной функции
(линейная регрессия)

\begin{equation*}
F\left( {x,a,b} \right) = ax + b.
\end{equation*}

Составим систему вида (1)
$$
\left\{ \begin{array}{l}
 {\sum\limits_{i = 0}^{n} {\left( {y_{i} - ax_{i} - b} \right) \cdot x_{i} =
0,}}  \\
 {\sum\limits_{i = 0}^{n} {\left( {y_{i} - ax_{i} - b} \right) = 0.}}  \\
 \end{array}  \right.
$$

Преобразуем эту систему к равносильной
$$
\left\{
\begin{array}{l}
 \left( \frac{1}{n}\sum\limits_{i=0}^n \left( x_i  \right)^2 \right)
 \cdot a + \left( \frac{1}{n}\sum\limits_{i=0}^n x_i  \right)
 \cdot b = \frac{1}{n}\sum\limits_{i=0}^n y_i x_i,\\
 \left( \frac{1}{n}\sum\limits_{i=0}^n x_i  \right) \cdot
 a + b = \frac{1}{n}\sum\limits_{i=0}^n y_i  ,\quad i = 1, \ldots ,n. \\
 \end{array} \right.
$$

Решая эту систему, найдем значения параметров $a$ и $b.$

Необходимым условием для выбора линейной функции в качестве искомой эмпирической является соотношение
$$y(\frac{x_1+x_n}{2})-\sqrt{y(x_1)y(x_n)}=0.$$

{\bf Квадратичная функция (квадратичная регрессия).}

Будем искать приближающую функцию в виде квадратного трехчлена.
$$
F(x,a,b,c)=ax^2+bx+c.
$$
Находим частные производные:
$$
F'_a=x^2, F'_b=x, F'_c=1.$$

Составим систему вида
$$
\left\{
\begin{array}{ll}
\sum\limits_i(y_i-a(x_i)^2-bx_i-c)\cdot (x_i)^2=0,\\
\sum\limits_i(y_i-a(x_i)^2-bx_i-c)\cdot x_i=0,\\
\sum\limits_i(y_i-a(x_i)^2-bx_i-c)=0, i=1,\dots ,n.\\
\end{array}
\right.
$$

После несложных преобразований получается система трех линейных уравнений с тремя неизвестными $a,b,c.$

Квадратичная регрессия применяется, если все выражения вида $y_2-2y_1+y_0, y_3-2y_2+y_1, y_4-2y_3+y_2$ и т.д. мало отличаются друг от друга.

Линейная регрессия (Linear regression)

$f(x) = ax + b$

```python
import numpy as np
x = np.linspace(0,10,20)
#X = np.array(x).reshape(-1, 1)
#print(X)
#y = [0, 2.2, 1.5, 2.7, 4.3, 4.7, 6.25, 6.68, 8.12, 9.17, 10.1]
y = [el + np.random.uniform(-1,1) for el in x]
#Y = np.array(y).reshape(-1, 1)
#print(Y)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.linspace(0, 10, 100)
y1 = x1
plt.plot(x, y, 'go')
plt.plot(x1, y1)
plt.show()
```

```python
import numpy as np
n = len(x)
xi = sum(x)/n
yi = sum(y)/n

def sumxy(x,y):
  s = 0
  for i in range(len(x)):
    s+=x[i]*y[i]
  return s

xiyi = sumxy(x,y)/n

xi2 = sum([el*el for el in x])/n

A = np.matrix([[xi2, xi], [xi, 1]])
B = np.matrix([[xiyi], [yi]])
res = np.linalg.solve(A, B)
print(res)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.linspace(0, 10, 100)
y1 = x1
y2 = float(res[0])*x1+ float(res[1])
plt.plot(x, y, 'go')
plt.plot(x1, y1, x1, y2)
plt.show()
```

```python
model = LinearRegression()

model.fit(np.array(x).reshape(-1, 1), np.array(y).reshape(-1, 1)) # обучение модели

a = model.coef_
b = model.intercept_
print(a, b)
```
