**Градиентный спуск** — итерационный метод оптимизации, используемый для минимизации функции потерь.

# Градиент

**Градиент** — это вектор из частных производных функции по всем переменным. Он показывает направление наибольшего роста функции в точке.

## Определение градиента

Пусть задана функция многих переменных:

$$
f(x_1, x_2, …, x_n)
$$

Её градиент вычисляется так:

$$
\nabla f(\vec{x}) = \left( \frac{\partial f}{\partial x_1},\, \frac{\partial f}{\partial x_2},\, …,\, \frac{\partial f}{\partial x_n} \right)
$$

- Каждый компонент — производная по соответствующей переменной.

## Интуиция

- Градиент указывает направление, в котором функция возрастает быстрее всего.
- В методе градиентного спуска движение идёт против градиента, чтобы найти минимум.

## Свойства

- В точке минимума: $\nabla f(\vec{x}_*) = 0$
- Если длина градиента близка к нулю ($||\nabla f(\vec{x}_*)|| < \varepsilon $), обычно процесс оптимизации останавливают.

## Основная идея

На каждом шаге параметры модели обновляются по формуле:

$$
\vec{x}_{k+1} = \vec{x}_k - \alpha \nabla f(\vec{x}_k)
$$

где
- $\vec{x}_k$ — текущие параметры,
- $\alpha$ — шаг обучения (learning rate),
- $\nabla f(\vec{x}_k)$ — градиент функции в данной точке.

## Пошаговый алгоритм

1. **Инициализация**: Задаём начальное приближение $\vec{x}_0.$
2. **Вычисляем градиент**: $\nabla f(\vec{x}_k).$
3. **Обновляем параметры**: $\vec{x}_{k+1} = \vec{x}_k - \alpha \nabla f(\vec{x}_k)$.
4. **Проверяем условие остановки**: Если градиент мал, процесс завершён. Иначе, возвращаемся к шагу 2.

## Пример применения

Минимизация квадратичной функции:

$$
f(x) = x^2 + 4x + 3
$$
$$
f'(x) = 2x + 4
$$

Итерации градиентного спуска:



```python
x = 5.0 # Начальное приближение
learning_rate = 0.01
for _ in range(100):
    grad = 2 * x + 4
    x -= learning_rate * grad
print("Итоговое значение:", x)
```

## Преимущества и недостатки

- Метод прост для реализации.
- Чувствителен к выбору шага $\alpha$: слишком большой ведёт к расходимости, слишком малый — к медленной сходимости.
- Для сложных функций с "оврагами" может сходиться медленно.



```python
import plotly.graph_objects as go
import numpy as np

# Функция и её производная
def f(x):
    return x**2 + 4*x + 4

def df(x):
    return 2*x + 4

# Параметры градиентного спуска
x0 = 3.0          # начальная точка
alpha = 0.05      # шаг обучения (learning rate)
max_iter = 50     # количество итераций

# Вычисляем траекторию градиентного спуска
trajectory_x = [x0]
trajectory_y = [f(x0)]

x_curr = x0
for i in range(max_iter):
    grad = df(x_curr)
    x_curr = x_curr - alpha * grad
    trajectory_x.append(x_curr)
    trajectory_y.append(f(x_curr))

# График функции
x_range = np.linspace(-8, 6, 200)
y_range = f(x_range)

fig = go.Figure()

# Основная функция
fig.add_trace(go.Scatter(
    x=x_range,
    y=y_range,
    mode='lines',
    name='f(x) = x² + 4x + 4',
    line=dict(color='blue', width=2)
))

# Текущая точка (красная)
fig.add_trace(go.Scatter(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    mode='markers',
    name='Текущая точка',
    marker=dict(size=12, color='red', line=dict(width=2, color='white'))
))

# Траектория (точки)
fig.add_trace(go.Scatter(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    mode='markers',
    name='Траектория',
    marker=dict(size=6, color='orange', opacity=0.6)
))

# Касательная линия
x_tangent = np.linspace(trajectory_x[0] - 2, trajectory_x[0] + 2, 50)
slope = df(trajectory_x[0])
y_tangent = trajectory_y[0] + slope * (x_tangent - trajectory_x[0])

fig.add_trace(go.Scatter(
    x=x_tangent,
    y=y_tangent,
    mode='lines',
    name='Касательная',
    line=dict(color='red', width=2, dash='dash')
))

# Создаём слайдер для итераций
steps = []
for i in range(len(trajectory_x)):
    # Касательная в текущей точке
    x_tan = np.linspace(trajectory_x[i] - 2, trajectory_x[i] + 2, 50)
    slope_i = df(trajectory_x[i])
    y_tan = trajectory_y[i] + slope_i * (x_tan - trajectory_x[i])

    step = dict(
        method='update',
        label=f'{i}',
        args=[
            {
                'x': [x_range, [trajectory_x[i]], trajectory_x[:i+1], x_tan],
                'y': [y_range, [trajectory_y[i]], trajectory_y[:i+1], y_tan]
            },
            {
                'title': f'Итерация {i}: x = {trajectory_x[i]:.4f}, f(x) = {trajectory_y[i]:.4f}, градиент = {slope_i:.4f}'
            }
        ]
    )
    steps.append(step)

slider = dict(
    active=0,
    currentvalue={"prefix": "Итерация: ", "visible": True},
    steps=steps,
    pad={"b": 10, "t": 50},
    len=0.9,
    x=0.05
)

fig.update_layout(
    sliders=[slider],
    title=f'Итерация 0: x = {trajectory_x[0]:.4f}, f(x) = {trajectory_y[0]:.4f}, градиент = {df(trajectory_x[0]):.4f}',
    xaxis_title='x',
    yaxis_title='f(x)',
    xaxis=dict(range=[-8, 6]),
    yaxis=dict(range=[-2, 30]),
    width=900,
    height=600,
    margin=dict(b=150),
    showlegend=True
)

fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# Функция двух переменных и её градиент
def f(x, y):
    return x**2 + y**2 + 2*x - 4*y + 5

def grad_x(x, y):
    return 2*x + 2

def grad_y(x, y):
    return 2*y - 4

# Параметры градиентного спуска
x0, y0 = 5.0, 5.0
alpha = 0.025
max_iter = 30

# Траектория
trajectory_x = [x0]
trajectory_y = [y0]
trajectory_z = [f(x0, y0)]

x_curr, y_curr = x0, y0
for i in range(max_iter):
    gx = grad_x(x_curr, y_curr)
    gy = grad_y(x_curr, y_curr)

    x_curr = x_curr - alpha * gx
    y_curr = y_curr - alpha * gy

    trajectory_x.append(x_curr)
    trajectory_y.append(y_curr)
    trajectory_z.append(f(x_curr, y_curr))

# Создаём сетку для поверхности
x_surf = np.linspace(-5, 5, 60)
y_surf = np.linspace(-5, 8, 60)
X_surf, Y_surf = np.meshgrid(x_surf, y_surf)
Z_surf = f(X_surf, Y_surf)

fig = go.Figure()

# Поверхность функции
fig.add_trace(go.Surface(
    x=X_surf, y=Y_surf, z=Z_surf,
    name='f(x,y)',
    colorscale='Viridis',
    opacity=0.85,
    showscale=True,
    colorbar=dict(title="f(x,y)", len=0.6, y=0.8)
))

# Все промежуточные точки траектории
fig.add_trace(go.Scatter3d(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    z=[trajectory_z[0]],
    mode='markers',
    name='Траектория',
    marker=dict(
        size=6,
        color=[0],
        colorscale='Plasma',
        showscale=True,
        colorbar=dict(title="Итерация", x=1.15, len=0.6, y=0.8),
        line=dict(width=1, color='white')
    )
))

# Линия траектории
fig.add_trace(go.Scatter3d(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    z=[trajectory_z[0]],
    mode='lines',
    name='Путь',
    line=dict(color='orange', width=4)
))

# Текущая точка
fig.add_trace(go.Scatter3d(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    z=[trajectory_z[0]],
    mode='markers',
    name='Текущая точка',
    marker=dict(size=10, color='red', symbol='diamond', line=dict(width=2, color='white'))
))

# Проекция траектории на дно
z_bottom = Z_surf.min() - 5
fig.add_trace(go.Scatter3d(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    z=[z_bottom] * 1,
    mode='markers+lines',
    name='Проекция',
    marker=dict(size=4, color='gray', opacity=0.5),
    line=dict(color='gray', width=2, dash='dot')
))

# Слайдер
steps = []
for i in range(len(trajectory_x)):
    step = dict(
        method='update',
        label=f'{i}',
        args=[
            {
                'x': [
                    X_surf,
                    trajectory_x[:i+1],
                    trajectory_x[:i+1],
                    [trajectory_x[i]],
                    trajectory_x[:i+1]
                ],
                'y': [
                    Y_surf,
                    trajectory_y[:i+1],
                    trajectory_y[:i+1],
                    [trajectory_y[i]],
                    trajectory_y[:i+1]
                ],
                'z': [
                    Z_surf,
                    trajectory_z[:i+1],
                    trajectory_z[:i+1],
                    [trajectory_z[i]],
                    [z_bottom] * (i+1)
                ],
                'marker.color': [
                    None,
                    list(range(i+1)),
                    None,
                    None,
                    None
                ]
            },
            {
                'title': f'Итерация {i}: x={trajectory_x[i]:.3f}, y={trajectory_y[i]:.3f}, f={trajectory_z[i]:.3f}'
            }
        ]
    )
    steps.append(step)

fig.update_layout(
    sliders=[dict(
        active=0,
        currentvalue={"prefix": "Итерация: "},
        steps=steps,
        pad={"b": 10, "t": 60},
        len=0.9,
        x=0.05
    )],
    title=f'Градиентный спуск (α={alpha}): x={trajectory_x[0]:.3f}, y={trajectory_y[0]:.3f}',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='f(x,y)',
        camera=dict(eye=dict(x=1.5, y=1.5, z=1.2))
    ),
    width=1000,
    height=800,
    margin=dict(b=150)
)

fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# Итерационная формула (например, метод градиентного спуска)
def iteration_step(x, y, learning_rate=0.1):
    """
    Пример: градиентный спуск для функции f(x,y) = x^2 + y^2
    Градиент: (2x, 2y)
    """
    grad_x = 2 * x
    grad_y = 2 * y

    new_x = x - learning_rate * grad_x
    new_y = y - learning_rate * grad_y

    return new_x, new_y

# Начальная точка
x0, y0 = 4.0, 3.0

# Вычисляем траекторию итераций
max_iterations = 50
trajectory_x = [x0]
trajectory_y = [y0]

x_curr, y_curr = x0, y0
for i in range(max_iterations):
    x_curr, y_curr = iteration_step(x_curr, y_curr)
    trajectory_x.append(x_curr)
    trajectory_y.append(y_curr)

# Создаём фигуру
fig = go.Figure()

# Начальная точка (только первая)
fig.add_trace(go.Scatter(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    mode='markers',
    name='Траектория',
    marker=dict(
        size=10,
        color='red',
        line=dict(width=2, color='white')
    )
))

# Линия траектории (начально пустая)
fig.add_trace(go.Scatter(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    mode='lines',
    name='Путь',
    line=dict(color='blue', width=2)
))

# Контурная карта функции для визуализации
x_grid = np.linspace(-5, 5, 100)
y_grid = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x_grid, y_grid)
Z = X**2 + Y**2

fig.add_trace(go.Contour(
    x=x_grid,
    y=y_grid,
    z=Z,
    colorscale='Viridis',
    opacity=0.3,
    showscale=False,
    contours=dict(showlabels=True)
))

# Создаём слайдер для итераций
steps = []
for i in range(len(trajectory_x)):
    step = dict(
        method='update',
        label=f'Итер. {i}',
        args=[
            {
                'x': [trajectory_x[:i+1], trajectory_x[:i+1], x_grid],
                'y': [trajectory_y[:i+1], trajectory_y[:i+1], y_grid]
            },
            {
                'title': f'Итерация {i}: x={trajectory_x[i]:.3f}, y={trajectory_y[i]:.3f}'
            }
        ]
    )
    steps.append(step)

slider = dict(
    active=0,
    currentvalue={"prefix": "Итерация: "},
    steps=steps,
    pad={"b": 10, "t": 50},
    len=0.9,
    x=0.05
)

fig.update_layout(
    sliders=[slider],
    title=f'Итерация 0: x={trajectory_x[0]:.3f}, y={trajectory_y[0]:.3f}',
    xaxis_title='X',
    yaxis_title='Y',
    xaxis=dict(range=[-5, 5]),
    yaxis=dict(range=[-5, 5]),
    width=900,
    height=700,
    margin=dict(b=150)
)

fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# Функция и градиент
def f(x, y):
    return (x - 1)**2 + (y + 2)**2 + 0.5*x*y

def grad_x(x, y):
    return 2*(x - 1) + 0.5*y

def grad_y(x, y):
    return 2*(y + 2) + 0.5*x

# Параметры
x0, y0 = 5.0, 4.0
alpha = 0.15
max_iter = 40

# Траектория
trajectory_x = [x0]
trajectory_y = [y0]

x_curr, y_curr = x0, y0
for i in range(max_iter):
    x_curr = x_curr - alpha * grad_x(x_curr, y_curr)
    y_curr = y_curr - alpha * grad_y(x_curr, y_curr)
    trajectory_x.append(x_curr)
    trajectory_y.append(y_curr)

# Сетка для контурной карты
x_grid = np.linspace(-2, 6, 100)
y_grid = np.linspace(-6, 5, 100)
X, Y = np.meshgrid(x_grid, y_grid)
Z = f(X, Y)

fig = go.Figure()

# Контурная карта
fig.add_trace(go.Contour(
    x=x_grid, y=y_grid, z=Z,
    colorscale='Viridis',
    opacity=0.7,
    showscale=True,
    colorbar=dict(title="f(x,y)"),
    contours=dict(
        showlabels=True,
        labelfont=dict(size=10)
    )
))

# Все промежуточные точки
fig.add_trace(go.Scatter(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    mode='markers',
    name='Траектория',
    marker=dict(
        size=10,
        color=[0],
        colorscale='Plasma',
        showscale=True,
        colorbar=dict(title="Итерация", x=1.15),
        line=dict(width=2, color='white')
    )
))

# Линия пути
fig.add_trace(go.Scatter(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    mode='lines',
    name='Путь',
    line=dict(color='orange', width=3)
))

# Текущая точка
fig.add_trace(go.Scatter(
    x=[trajectory_x[0]],
    y=[trajectory_y[0]],
    mode='markers',
    name='Текущая',
    marker=dict(size=15, color='red', symbol='star', line=dict(width=2, color='white'))
))

# Стрелка градиента в текущей точке
gx0 = grad_x(trajectory_x[0], trajectory_y[0])
gy0 = grad_y(trajectory_x[0], trajectory_y[0])
fig.add_trace(go.Scatter(
    x=[trajectory_x[0], trajectory_x[0] - 0.5*gx0],
    y=[trajectory_y[0], trajectory_y[0] - 0.5*gy0],
    mode='lines',
    name='Градиент',
    line=dict(color='red', width=3),
    showlegend=True
))

# Слайдер
steps = []
for i in range(len(trajectory_x)):
    gx_i = grad_x(trajectory_x[i], trajectory_y[i])
    gy_i = grad_y(trajectory_x[i], trajectory_y[i])

    step = dict(
        method='update',
        label=f'{i}',
        args=[
            {
                'x': [
                    x_grid,
                    trajectory_x[:i+1],
                    trajectory_x[:i+1],
                    [trajectory_x[i]],
                    [trajectory_x[i], trajectory_x[i] - 0.5*gx_i]
                ],
                'y': [
                    y_grid,
                    trajectory_y[:i+1],
                    trajectory_y[:i+1],
                    [trajectory_y[i]],
                    [trajectory_y[i], trajectory_y[i] - 0.5*gy_i]
                ],
                'marker.color': [None, list(range(i+1)), None, None, None]
            },
            {
                'title': f'Итерация {i}: x={trajectory_x[i]:.3f}, y={trajectory_y[i]:.3f}, f={f(trajectory_x[i], trajectory_y[i]):.3f}'
            }
        ]
    )
    steps.append(step)

fig.update_layout(
    sliders=[dict(
        active=0,
        currentvalue={"prefix": "Итерация: "},
        steps=steps,
        pad={"b": 10, "t": 50},
        len=0.9,
        x=0.05
    )],
    title='Градиентный спуск (вид сверху)',
    xaxis_title='X',
    yaxis_title='Y',
    width=900,
    height=800,
    margin=dict(b=150)
)

fig.show()
```
