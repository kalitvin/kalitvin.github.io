Вычисление интеграла методом Монте-Карло

```python
import numpy as np

rng = np.random.default_rng(42)

def f(x):
    return np.exp(-x**2)

def monte_carlo_integral(f, a=0.0, b=1.0, n_samples=10_000):
    # 1. генерируем равномерные точки на [a, b]
    x = rng.uniform(a, b, size=n_samples)
    # 2. считаем значения функции
    fx = f(x)
    # 3. оценка интеграла: (b - a) * среднее значение
    I_hat = (b - a) * fx.mean()
    # 4. оценка стандартной ошибки
    s2 = fx.var(ddof=1)
    se = (b - a) * np.sqrt(s2 / n_samples)
    return I_hat, se

I_est, I_se = monte_carlo_integral(f, a=0.0, b=1.0, n_samples=100_000)

print(f"Оценка интеграла: {I_est:.6f}")
print(f"Стандартная ошибка: {I_se:.6f}")
```
